## Part 1 — MGRS Alignment

Sentinel-2 images are delivered in **MGRS tiles**, each with a fixed
pixel grid anchored at the tile's upper-left corner.
To extract a patch that lines up perfectly with those pixels we need to
snap our centroid-based window to the nearest valid grid position.

```
  centroid (lon/lat)
       │
       ▼
  UTM easting/northing
       │
       ▼
  raw upper-left corner  (centroid − half_patch)
       │
       ▼
  snapped upper-left     (aligned to 60 m grid anchored at MGRS origin)
```

The snapping interval is **60 m** — the LCM of Sentinel-2 resolutions
(10, 20, 60 m) — so the window is valid at every resolution.

````python
# ── Load grid points (100k sample from MajorTOM) ──────────────────────────
GRID_URL = "https://huggingface.co/datasets/JulioContrerasH/MJv2/resolve/main/MT_grid_10km_sample_100k.parquet"
MGRS_URL = "https://huggingface.co/datasets/JulioContrerasH/MJv2/resolve/main/mgrs_grid.parquet"

grid = pd.read_parquet(GRID_URL)
mgrs_df = pd.read_parquet(MGRS_URL)
mgrs = gpd.GeoDataFrame(
    mgrs_df,
    geometry=gpd.GeoSeries.from_wkb(mgrs_df["geometry"]),
    crs="EPSG:4326",
)

print(f"Grid points : {len(grid):,}")
print(f"MGRS tiles  : {len(mgrs):,}")
grid.head(3)
````

### Example 1 — Single point (step by step)

Before running the whole table it helps to trace one point manually
so we can see exactly what each step produces.

````python
# Pick one point
point = grid.iloc[0]
lon, lat = point["centre_lon"], point["centre_lat"]
utm_crs  = point["utm_crs"]

print(f"Grid cell : {point['grid_cell']}")
print(f"Lon / Lat : {lon:.4f}  {lat:.4f}")
print(f"UTM CRS   : {utm_crs}")

# Step 1 — find which MGRS tile contains this point
pt_gdf = gpd.GeoDataFrame(
    [{"grid_cell": point["grid_cell"]}],
    geometry=gpd.points_from_xy([lon], [lat]),
    crs="EPSG:4326",
)
hit = gpd.sjoin(pt_gdf, mgrs[["Name", "epsg", "ul_x", "ul_y", "geometry"]],
                how="left", predicate="intersects")
tile = hit[hit["epsg"] == int(utm_crs.split(":")[1])].iloc[0]

print(f"\nMGRS tile  : {tile['Name']}")
print(f"Tile origin: ul_x={tile['ul_x']}  ul_y={tile['ul_y']}")

# Step 2 — convert centroid to UTM
east, north = centroid_to_utm(lon, lat, utm_crs)
print(f"\nUTM easting / northing : {east:.1f}  {north:.1f}")

# Step 3 — compute raw upper-left (centroid minus half the patch)
PATCH_SIZE, PIXEL_SIZE = 1024, 10
half     = PATCH_SIZE * PIXEL_SIZE / 2          # 5120 m
raw_tx   = east  - half
raw_ty   = north + half
print(f"Raw upper-left         : tx={raw_tx:.1f}  ty={raw_ty:.1f}")

# Step 4 — snap to grid
tx, ty = snap_to_grid(raw_tx, raw_ty, tile["ul_x"], tile["ul_y"], grid_align=60)
print(f"Snapped upper-left     : tx={tx:.1f}  ty={ty:.1f}")

# Verify alignment
print(f"\nAlignment check:")
print(f"  (tx - ul_x) % 60 = {(tx - tile['ul_x']) % 60}  (should be 0)")
print(f"  (ul_y - ty) % 60 = {(tile['ul_y'] - ty) % 60}  (should be 0)")
````

The same four steps above are wrapped in `compute_patch_origin`:

````python
tx2, ty2 = compute_patch_origin(
    lon=lon, lat=lat, utm_crs=utm_crs,
    ul_x=tile["ul_x"], ul_y=tile["ul_y"],
    patch_size=1024, pixel_size=10, grid_align=60,
)
print(f"compute_patch_origin → tx={tx2:.1f}  ty={ty2:.1f}")
assert tx == tx2 and ty == ty2, "mismatch!"
print("Matches manual calculation ✓")
````

### Example 2 — Small batch (100 points)

`build_patch_transforms` handles the full table: spatial join,
UTM conversion, and snapping — all in parallel using subprocesses.
Here we run it on 100 random points as a quick sanity check.

````python
sample = grid.sample(n=100, random_state=42).reset_index(drop=True)

result = build_patch_transforms(
    df         = sample,
    mgrs       = mgrs,
    n_workers  = 1,        # single process for a quick test
    chunk_size = 100,
)

print(f"\nOutput: {len(result)} rows")
result[["grid_cell", "mgrs_tile", "utm_crs", "translateX", "translateY"]].head(10)
````

### Example 3 — Full table

Same call, but now on the complete 100k grid with all available CPU cores.
The result is saved to parquet and reused in Part 2.

````python
grid_aligned = build_patch_transforms(
    df         = grid,
    mgrs       = mgrs,
    n_workers  = 8,
    chunk_size = 25_000,
)

grid_aligned.to_parquet("grid_aligned.parquet", index=False)
print(f"Saved {len(grid_aligned):,} rows → grid_aligned.parquet")
grid_aligned.head(3)
````

## Part 2 — Best Image Selection

For each grid point we need to pick **one satellite image** to download.
Good criteria: high cloud-free score, spread across all four seasons,
and spread across years so the dataset isn't all from a single date.

The workflow has two steps:

```
  run()              — query GEE for every point, score each image
       │                 returns top-N candidates per point
       ▼
  seasonal_select()  — pick the single best image per point
                        balancing seasons and years globally
```

````python
import ee
ee.Authenticate()
ee.Initialize(project="ee-julius013199")
````

### Sensor configuration

A `SensorConfig` tells `run()` which GEE collection to query,
how to compute the clear-pixel score, and how many candidates to keep.
We use the pre-built `LANDSAT_LC08_L2` config and override only what we need.

````python
def lc08_clear_mask(img: ee.Image) -> ee.Image:
    """Clear if bit-6 (clear) is set AND bit-4 (shadow) is not."""
    qa     = img.select("QA_PIXEL")
    clear  = qa.bitwiseAnd(1 << 6).neq(0)
    shadow = qa.bitwiseAnd(1 << 4).neq(0)
    return clear.And(shadow.Not()).rename("clear").toFloat()


cfg = SensorConfig(
    name        = "LC08_L2",
    collections = ["LANDSAT/LC08/C02/T1_L2", "LANDSAT/LC08/C02/T2_L2"],
    quality     = QualityConfig(
        mask_fn  = lc08_clear_mask,
        scale    = 30,
        tile_px  = 400,
        reducer  = "mean",
    ),
    date_range  = ("2013-04-11", "2026-03-13"),
    top_n       = 10,     # keep 10 candidates per point; seasonal_select picks 1
)

print(cfg)
````

### Example 1 — Single point

A minimal test before running the full table.
We use one point over central Peru where Landsat coverage is well known.

````python
one_point = pd.DataFrame([{
    "grid_cell"  : "test_peru",
    "centre_lon" : -74.0,
    "centre_lat" : -12.0,
}])

# Score every available image at this point
candidates = run(one_point, cfg, method="reduceregion", n_workers=4)
print(f"Found {len(candidates)} candidates")
candidates.sort_values("score", ascending=False).head(10)
````

````python
# Pick the best one, respecting season balance
final = seasonal_select(
    df_candidates = candidates,
    df_meta       = one_point,
    year_range    = (2013, 2026),
    spread_years  = True,
)
final[["grid_cell", "id_gee", "score", "date", "season_assigned"]]
````

### Example 2 — Spatial subset (quick production test)

Before committing to the full 100k table, we test on a small spatial window
to verify the pipeline end-to-end and check the season/year distributions.

````python
# ~50-100 points around central Peru
subset = grid_aligned[
    grid_aligned["centre_lon"].between(-76, -74) &
    grid_aligned["centre_lat"].between(-13, -11)
].reset_index(drop=True)

print(f"{len(subset)} points in spatial subset")

# Score
candidates_sub = run(
    subset, cfg,
    method    = "reduceregion",
    n_workers = 20,
    verbose   = True,
)
candidates_sub.to_parquet("candidates_subset.parquet", index=False)

# Select
final_sub = seasonal_select(
    df_candidates  = candidates_sub,
    df_meta        = subset,
    year_range     = (2013, 2026),
    balance_globally = True,
    spread_years   = True,
)
final_sub.to_parquet("final_subset.parquet", index=False)

final_sub[["grid_cell", "id_gee", "score", "date", "season_assigned"]].head(15)
````

### Example 3 — Full table (100k points)

Production run on the complete aligned grid.
This is the file that feeds Part 3 (patch download).

````python
# Score — the slow part (~hours for 100k points)
candidates_full = run(
    grid_aligned, cfg,
    method    = "reduceregion",
    n_workers = 40,
)
candidates_full.to_parquet("candidates_lc08_full.parquet", index=False)
print(f"Candidates saved: {len(candidates_full):,} rows")
````

````python
# Select — fast
final_full = seasonal_select(
    df_candidates    = candidates_full,
    df_meta          = grid_aligned,
    year_range       = (2013, 2026),
    balance_globally = True,
    spread_years     = True,
)
final_full.to_parquet("final_lc08_full.parquet", index=False)

# Sanity checks
assert final_full["grid_cell"].nunique() == len(final_full), "duplicate grid cells!"
na_count = (final_full["id_gee"] == "NA").sum()
print(f"Total points      : {len(final_full):,}")
print(f"With valid image  : {len(final_full) - na_count:,}")
print(f"No image found    : {na_count:,}")

final_full[["grid_cell", "id_gee", "score", "date", "season_assigned"]].head(10)
````
